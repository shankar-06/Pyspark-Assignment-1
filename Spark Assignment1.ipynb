### Spark Assignment 1

#### Word Count End to End
1. Create the source RDD
Two lines of text.

2. flatMap: split lines into words
6 words are produced (flat).

3. map: turn each word into (word, 1)
Creates a pair RDD.

4. reduceByKey: add the 1s per word
Wide transformation – shuffle happens here.

5. collect: the action
Triggers all the steps above and returns the result.



```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder\
                    .appName('Assignment 1')\
                    .master('local[*]')\
                    .getOrCreate()
```

```python
sc = spark.sparkContext
```

#### Question 1

```python
rdd = sc.parallelize(['Welcome To Spark', 'Welcome To Snowflake'])
```

#### Question 2

```python
fm = rdd.flatMap(lambda x:x.split(' '))
```

#### Question 3

```python
m = fm.map(lambda x:(x,1))
```

#### Question 4

```python
rbk = m.reduceByKey(lambda x,y:x+y)
```

#### Question 5

```python
rbk.collect()
```
